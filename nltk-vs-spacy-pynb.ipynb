# **SPACY**


```python
import spacy 
spacy.load("en_core_web_sm")
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Dr. Abir is a bit angry today. Seems like he has lost his mind")
#doc.sentiment
for sentence in doc.sents:
    print(sentence)
```

```python
for semtence in doc.sents:
    for word in sentence:
        print(word)     # print words as token.
```

# **nltk**

```python
import nltk
nltk.download("punkt_tab")
```

```python
from nltk.tokenize import sent_tokenize
sentences=sent_tokenize("Dr. Abir is a bit angry today. Seems like he has lost his mind")
sentences
```

```python
from nltk.tokenize import word_tokenize
Words=word_tokenize("Dr. Abir is a bit angry today. Seems like he has lost his mind")
Words
```
